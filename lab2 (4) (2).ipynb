## Instrukcja match ... case

```python
"""
match zmienna:
    case wartość1:
        instrukcje1
    case wartość2:
        instrukcje2
    ...
    case _: # (opcjonalnie) znak _ oznacza tutaj 'cokolwiek' 
        intrukcja_defaultowa

"""
```

```python
R = int(input("Podaj numer błędu: "))

match R:
    case 400:
        print("Błąd 400: Nieprawidłowy żądanie")
    case 404:
        print("Błąd 404: Nie znaleziono zasobu")
    case 500:
        print("Błąd 500: Wewnętrzny błąd serwera")
    case _:
        print("Nieznany błąd")
```

# Definiowanie funkcji

```python
"""
def nazwa_funkcji(parametry):
    instrukcje
    return wartość
"""
```

```python
def policz_pole_prostokata(a, b):
    return a * b

def funkcja_kwadratowa(a, b, c):
    delta = b**2 - 4*a*c
    if delta < 0:
        return "Brak pierwiastków rzeczywistych"
    elif delta == 0:
        x = -b / (2*a)
        return f"Jeden pierwiastek: {x}"
    else:
        x1 = (-b + delta**0.5) / (2*a)
        x2 = (-b - delta**0.5) / (2*a)
        return f"Dwa pierwiastki: {x1} i {x2}"
```

### Wywoływanie funkcji

```python
"""
wynik = nazwa_funkcji(argumenty)
"""
```

```python
print(policz_pole_prostokata(5, 10))

print(funkcja_kwadratowa(1, -3, 2))
```

```python
print(policz_pole_prostokata(a=5, b=10))
print(policz_pole_prostokata(b=10, a=5))
print(policz_pole_prostokata(5, b=10))
```

Instrukcja while będzie wykonywać kod dopóki warunek jest prawdziwy. Uwaga na nieskończone pętle - należy w instrukcjach zawrzeć w jakiś sposób zmianę w warunku albo instrukcję przerwania pętli

```python
for i in range(1, 6):
    print(f"Iteracja {i}")
```

```python
for i in range(1, 11, 2):
    print(f"Iteracja {i}")
```

```python
while i < 5:
    print(f"Iteracja {i}")
    i += 1
```

### Zatrzymanie pętli - break

```python
for i in range(1, 6):
    print(f"Iteracja {i}")
    if i == 3:
        break
```

### Pominięcie wykonywania reszty instrukcji

```python
for i in range(5):
    if i == 3:
        continue
    print(f"Iteracja {i}")
```
